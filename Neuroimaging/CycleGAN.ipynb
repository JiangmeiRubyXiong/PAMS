```python
import numpy as np
import nibabel as nib
import os

# Directory containing slices for all subjects
slice_dir = "/media/disk2/beijing_dti/Neuroimaging/train_T1_FA/synthetic_1mm/"  # Update this to your actual folder

# Assuming slices are named like "slice_001.nii.gz", "slice_002.nii.gz", ..., for each subject
n_slices = 182
n_subjects = 89  # Update to the actual number of subjects

# Prepare to store reassembled brains for all subjects
brains = []

for subj in range(n_subjects):
    # Initialize a list to hold slices for the current subject
    subject_slices = []

    for z in range(n_slices):
        # Load the z-th slice for this subject
        
        slice_path = os.path.join(slice_dir, f"synthetic{z}/FA_synthetic_epoch_200.nii.gz")
        slice_data = nib.load(slice_path).get_fdata()[:, :, 0, subj]  # [X, Y], drop singleton dim

        # Append the slice to the list
        subject_slices.append(slice_data)
        
        print("sub"+str(subj)+"slice"+str(z))

    # Stack all slices along the z-axis to form a 3D volume [X, Y, Z]
    brain_volume = np.stack(subject_slices, axis=2)
    brains.append(brain_volume)

# Optionally save each subject's brain volume to NIfTI
for subj_idx, brain in enumerate(brains):
    output_path = "/media/disk2/beijing_dti/Neuroimaging/train_T1_FA/synthetic_brain/"+f"subject_{subj_idx}_reassembled_brain.nii.gz"
    nib.save(nib.Nifti1Image(brain, affine=np.eye(4)), output_path)
    print(f"Saved: {output_path}")
```
